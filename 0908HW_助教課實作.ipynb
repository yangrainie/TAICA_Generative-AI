matplotlib使用直線將給的點一段一段連起來，試著把每個點畫出來(描點法)

```python
import numpy as np
import matplotlib.pyplot as plt

# 1.定義數學函數
def f(x):
    return np.sin(x) + 2

# 2. 定義x的範圍
x5 = np.linspace(-15, 15, 100)
y = f(x5) # 注意：這裡使用了x5作為參數

# 3.設定圖表
plt.figure(figsize=(10, 6)) #生成一個圖形figure，且大小設定為寬度10，高度6
plt.plot(x5, y, label='f(x) = sin(x) + 2 (Line Plot)', color='blue') # 繪製線條圖（line plot）。它以 x5 作為 x 軸座標，y 作為 y 軸座標。color='blue' 線條顏色設為藍色。
plt.scatter(x5, y, label='f(x) = sin(x) + 2 (Points Plot)', color='red', s=10) # 繪製描點法，與上一行說明類似
plt.title('Function Plot with Line and Points') #圖表標題
plt.xlabel('x') # x軸標籤
plt.ylabel('f(x)')# y軸標籤
plt.legend() #顯示圖例。將前面設定的plot、scatter樣式顯示出來
plt.grid(True) #在圖形中顯示網格grid
plt.show()
```
