## 3.畫一個函數圖形
- 如何正確呈現函數式子?
> - 使用單獨一個錢字符號: $f(x) = \sin(x) + 2$
> - 使用兩個錢字符號，讓函式單獨一行出現: $$f(x) = \sin(x) + 2$$

- 想畫 $f(x) = \sin(x) + 2$ , $θ = x / 10$ 的圖形

因為我想畫比較特別的函數圖形，所以上網找了花瓣圖形的

與GPT討論的結果是
原本課堂上的程式碼是屬於畫直角座標圖，但我想畫的圖形(花瓣)是極座標圖
所以照原本程式碼會畫不出來


```python
# X 為-5到5的區間，取100個點
#x = np.linspace(-5, 5, 100)

#如果 x 只有 -5 到 5，則 θ 只有 -0.5 到 0.5，角度轉得太少，所以畫不出花瓣。

# X 的範圍要夠大，所以使用pi(數學圓周率)，以覆蓋足夠多的角度
x = np.linspace(-10*np.pi, 10*np.pi, 1000) #區間為-31.4159到31.4159之間
```

```python
y = np.sin(x) + 2
theta = x / 10
```

```python
#plt.plot(x, y) 是畫直角座標 (水平位置, 垂直位置)
plt.polar(theta, y) #是畫極座標圖 (角度, 半徑)
```
