## 2.2 輸入資料格式的整理


全連接神經網路的輸入通常以一維向量表示，因此將 28×28 的圖片攤平(flatten)成 784 個像素值，也就是python裡numpy()裡的reshape，達到改變輸入資料的格式(形狀)。

![示意圖](https://d1dwq032kyr03c.cloudfront.net/upload/images/20220930/20150622NbVWr6cbcf.jpg)
[參考資料 Day 15. 深度學習模型 - CNN（一）](https://ithelp.ithome.com.tw/articles/10301321)

````python
x_train = x_train.reshape(60000, 784)/255
x_test = x_test.reshape(10000, 784)/255
````


* 總共有60000張圖片，每張圖片大小為28×28像素
* 28×28 = 784，所以每張圖片有784個像素值
* 每個像素值代表顏色深淺，原本範圍為0～255
* 除以255後變成0～1，也稱作正規化，有助於訓練模型

MNIST 有 0～9 共 10 個類別，因此將答案轉換成 10 個位置的 One-Hot Encoding，用其中一個位置的 1 表示正確類別，方便神經網路進行分類

## 3. 打造第一個神經網路

根據前面的輸入資料，每個訓練資料都是28*28矩陣，並且希望輸出結果為10個類別，故函數表達式為

$$\hat{f} \colon \mathbb{R}^{784} \to \mathbb{R}^{10}$$
$$\hat{f} \colon \mathbb{input}  \to \mathbb{output}$$

函數表達式解釋:
>* $\hat{f}$ 加上一個帽子^，表示這個函數是透過神經網路學習出來的函數。
>* $\mathbb{R}^{784}$ 表示神經網路的**input**為784個數值，也就是每個訓練資料28*28矩陣。
>* $\mathbb{R}^{10}$ 表示神經網路的**output**為10個數值，也就是最後產生的10個類別。


### 3.1 決定神經網路架構、讀入相關套件

目前先使用 ReLU 作為激發函數。由於輸入輸出先前的函數已經確定，因此這個階段在設計神經網路架構時，主要需要決定隱藏層的數量以及每層神經元的個數。


### 3.2 開始建構神經網路





````python
# 將打造的神經網路模型稱為model(可以自己取)
# sequential是Keras提供的一種神經網路模型架構，特色為一層接著一層排列，資料按照順序通過每一層。
model = Sequential()
````

####3.2.1. activation function的目的:
由於類神經網路在訓練時，會以線性的方式進行訓練，但實際的問題通常為非線性的，為了減少線性程度(使得類神經網路的結果output更準確)，便加入activation function，以增加訓練時的非線性(複雜)程度。

####3.2.2 常見的activation function:
>

* `ReLU`:輸入為正數即輸出原值，否則輸出0。最常用的激勵函數之一，因訓練速度較快。
$$f(x) = max(0,x)$$

* `Sigmoid`:適用於二元分類。
* `Tanh`:某些需要輸出正負值的模型。
* `Softmax`:適用於多元分類。在計算各種類別發生的機率時，所有機率分佈的總和會等於1。






````python
model.add(Dense(20, input_dim=784, activation='relu'))
#Dense代表全連結的神經網路
#Dense(這層隱藏層要放幾個神經元，輸入為幾個數字(幾個特徵)? ,activation function: relu、sigmoid、tanh等)
````

* 前面的ReLU，負責讓隱藏層學習複雜的非線性特徵
* 輸出層的activation換成softmax是因為要讓輸出類別的機率分佈總和等於1
* 若輸出層也使用ReLU，輸出層會被限制為正數(非負數)，會導致10個類別的機率分佈加總不等於1，因此改成Softmax。

````python
model.add(Dense(10, activation='softmax'))
````

### 3.3 組裝神經網路

####3.3.1 設定學習規則

使用 `compile` 才正式把神經網路建好。

* 決定 optimizer, 使用標準的 SGD

* loss損失函數:模型目前錯的狀況如何?
> * 模型預測結果與實際正確答案差多少?
> * mse(均方誤差):相減平方加起來不要開更號，平均誤差為多少?
* optimizer:如何根據錯誤調整權重?
> * SGD(Stochastic Gradient Decent)隨機梯度下降:一次跑一個樣本或是小批次(mini-batch)樣本，算出一次梯度或是小批次梯度的平均後就更新一次。
> * learning_rate可想像為模型每次更新權重的幅度。
> * 梯度是 Loss 對權重的變化率，可以用梯度的正負判斷權重變化對 Loss 的影響方向，用梯度的大小表示 Loss 對該權重變化的敏感程度。

* metrics:如何衡量模型好壞?
> * 作為模型表現的評估指標。
>> *  acc代表accuracy(準確率):全部情況中，預測正確的情況佔多少?
>> * 從前面可得知，測試資料有10000張MNIST 手寫數字圖片。
>>> 假設模型預測正確9500張、預測錯誤500張
>>>> $Accuracy = \frac{9500}{10000}$ =0.95

**若不加上`axis=-1`，會發生什麼變化?**

假設同時對3筆資料進行測試，正常情況下，每筆測試資料都會有各自的一排分數，如下表格所示，也就代表會找出3個最大值。
|        | 類別 0 | 類別 1 | 類別 2 |
|--------|--------|--------|--------|
| 圖片 1 | 0.1    | 0.7    | 0.2    |
| 圖片 2 | 0.8    | 0.1    | 0.1    |
| 圖片 3 | 0.2    | 0.3    | 0.5    |

但去除掉`axis=-1`，會被預設成從全部的測試結果裡找出一個最大值。
故`axis=-1`的作用在於根據每筆資料的分數去做比較。


由於原本輸入資料的格式為28x28矩陣，中間經過格是整理，已經攤平成一維向量(784維的向量)。測試資料後，為了要讓輸出結果能夠完整呈現原本測試圖片，故需要再`reshape`成28x28矩陣。

### 7. 用 Gradio 來展示

````python
def resize_image(inp):
    # 圖在 inp["layers"][0]
    image = np.array(inp["layers"][0], dtype=np.float32)
    image = image.astype(np.uint8)

    # 轉成 PIL 格式
    image_pil = Image.fromarray(image)

    # Alpha 通道設為白色, 再把圖從 RGBA 轉成 RGB
    background = Image.new("RGB", image_pil.size, (255, 255, 255))
    background.paste(image_pil, mask=image_pil.split()[3]) # 把圖片粘貼到白色背景上，使用透明通道作為遮罩
    image_pil = background

    # 轉換為灰階圖像
    image_gray = image_pil.convert("L")

    # 將灰階圖像縮放到 28x28, 轉回 numpy array
    img_array = np.array(image_gray.resize((28, 28), resample=Image.LANCZOS))

    # 配合 MNIST 數據集
    img_array = 255 - img_array

    # 拉平並縮放
    img_array = img_array.reshape(1, 784) / 255.0

    return img_array
````

````python
def recognize_digit(inp):
    img_array = resize_image(inp)
    prediction = model.predict(img_array).flatten()
    labels = list('0123456789')
    return {labels[i]: float(prediction[i]) for i in range(10)}
````

````python
iface = gr.Interface(
    fn=recognize_digit,
    inputs=gr.Sketchpad(),
    outputs=gr.Label(num_top_classes=3),
    title="MNIST 手寫辨識",
    description="請在畫板上繪製數字"
)

iface.launch(share=True, debug=True)
````
